```python
from pydantic import BaseModel
from typing import Dict, Any, Optional
from enum import Enum
```

```python
class GradeType(Enum):
    S = 'S'
    A = 'A'
    B = 'B'
    C = 'C'
    D = 'D'
    F = 'F'
    
class ProgressScore(BaseModel):
    score: float
    desc: str

class ProgressResult(BaseModel):
    grade: GradeType = GradeType.F         # Grade for points
    average_score: float = 0.0         # point ( 0 ~ 100)
    top_percent: float = 0.0           # Grades are in the top few percen
    scores: list[ProgressScore] = []  # detailed scores
```

```python
json_string1 = """{
        "grade": "A",
        "average_score": 93,
        "top_percent": null,
        "scores": [
            {
                "score": 30,
                "desc": "Stage clear score"
            },
            {
                "score": 63,
                "desc": "Number of failures score"
            }
        ]
}"""
json_string2 = """{
        "scores": [
            {
                "score": 30,
                "desc": "Stage clear score"
            },
            {
                "score": 63,
                "desc": "Number of failures score"
            }
        ]
}"""
```

```python
progress = ProgressResult.model_validate_json("{}")
```
