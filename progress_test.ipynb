```python
import requests,json
from db.model.progress import ProgressResponse
from pydantic import BaseModel, ConfigDict

BASE_URL = "http://localhost:8104/"
```

```python
def login():
    api = f"{BASE_URL}/users/token"
    body = {
        "username":"klingo01",
        "password":"klingo"
    }
    response = requests.post(api, data=body)
    return response.json()["access_token"]
def sample_scenario(token,scenario_id,stage_type,level):
    ## 인터뷰 데이터 가져오기(랜덤하게 5게)
    api = f"{BASE_URL}/scenario/stages/redis/{scenario_id}/{stage_type}/{level}"
    header = {
        "Authorization":f"Bearer {token}"
    }
    print(header)
    response = requests.get(api, headers=header)
    print(response.status_code)
    return response.json()
```

```python
## 시나리오 가져오기
token = login()
print(token)
scenario = sample_scenario(token,1,1,1)
scenario
```
