## Redis Test

```python
from redis import Redis

REDIS_HOST = '100.100.53.32'
REDIS_PORT = 6379
```

```python
r = Redis(host=REDIS_HOST, port=REDIS_PORT, decode_responses=True)
```

```python
r.hset('user-session:123', mapping={
    'name': 'John',
    "surname": 'Smith',
    "company": 'Redis',
    "age": 29
})
```

```python
r.hget('user-session:123','name')
```

```python
r.hgetall('user-session:123')
```
