```python
class SingleTone:
    _instance = None
    # my_object = None
    def __new__(cls):
        ## instance 생성용
        print("__new__")
        if cls._instance is None:
            cls._instance = super().__new__(cls)
            cls.my_object = None
            print(cls)
        return cls._instance
    
    def __init__(self) -> None:
        print(f"__init__ : {self}")
        if self.my_object is None:
            self.my_object = 'single tone'

class Test:
    def __init__(self) -> None:
        self.my_object = None
        print("__init__")
        if self.my_object is None:
            self.my_object = 'test class'
```
